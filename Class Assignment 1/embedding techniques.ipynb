# Embedding Techniques

```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from gensim.models import Word2Vec

corpus = [
    "the cat sat on the mat",
    "the dog barked at the cat",
    "the cat chased the mouse",
    "the dog chased the cat",
    "a mouse ran under the mat",
    "the dog and the cat are friends",
    "the mouse hid from the cat",
    "the cat and the dog slept on the mat",
]
```

## One-Hot Encoding

```python
# one-hot encoding
vocab = sorted(set(word for sentence in corpus for word in sentence.split()))
word_index = {word: idx for idx, word in enumerate(vocab)}

def one_hot(word):
    vector = np.zeros(len(vocab), dtype=int)
    vector[word_index[word]] = 1
    return vector

sentence = corpus[0].split()
one_hot_matrix = np.array([one_hot(word) for word in sentence])
pd.DataFrame(one_hot_matrix, index=sentence, columns=vocab)
```

## Bag of Words

```python
# bag of words
bow_vectorizer = CountVectorizer()
bow_matrix = bow_vectorizer.fit_transform(corpus)
pd.DataFrame(bow_matrix.toarray(), columns=bow_vectorizer.get_feature_names_out())
```

## TF-IDF

```python
# tf-idf
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)
pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out()).round(3)
```

## Word2Vec

```python
# word2vec
tokenized_corpus = [sentence.split() for sentence in corpus]
w2v_model = Word2Vec(tokenized_corpus, vector_size=50, window=3, min_count=1, sg=1)

cat_vector = w2v_model.wv["cat"]
similar_to_cat = w2v_model.wv.most_similar("cat")

cat_vector, similar_to_cat
```
